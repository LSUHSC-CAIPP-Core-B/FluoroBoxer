# Visualize Bounding Boxes on Random Dataset Images

This notebook applies YOLO bounding boxes to randomly selected images from the dataset.

## Overview
The notebook:
1. Loads randomly selected images from the dataset
2. Reads YOLO format label annotations
3. Draws bounding boxes on the images
4. Saves the annotated images for visualization and verification

```python
# Import Required Libraries
import cv2
import os
import random
from config import *
from CellProcessor import read_yolo_labels, yolo_to_original
```

## Notes
- The output images are saved with the prefix "Necroptosis_"
- White rectangles represent bounding boxes for detected cells
- To visualize different dataset splits, modify the `SPLIT` variable (options: "train/", "val/", "test/")
- Adjust `num_images_to_save` to control how many random images are processed

```python
# Configure Dataset Paths and Parameters
PATH_TO_DATASET = "Data/final_Data_set"
SPLIT = "test/"  # Options: "train/", "val/", "test/"
PATH_TO_DATASET_IMAGES = os.path.join(PATH_TO_DATASET, "images", SPLIT)
PATH_TO_DATASET_LABELS = os.path.join(PATH_TO_DATASET, "labels", SPLIT)
OUTPUT_PATH = "Data/final_Data_set/example_boxes_vizulazition/"
os.makedirs(OUTPUT_PATH, exist_ok=True)

print(f"Images path: {PATH_TO_DATASET_IMAGES}")
print(f"Labels path: {PATH_TO_DATASET_LABELS}")
print(f"Output path: {OUTPUT_PATH}")
```

```python
# Process Random Images and Draw Bounding Boxes
# Get list of images and select random samples
images = os.listdir(PATH_TO_DATASET_IMAGES)
num_images = len(images)
num_images_to_save = 10

# Randomly select image indices
images_to_save_ids = random.sample(range(num_images), min(num_images_to_save, num_images))

print(f"Total images: {num_images}")
print(f"Processing {len(images_to_save_ids)} random images...")

# Process each selected image
for image_to_save_id in images_to_save_ids:
    image_name = images[image_to_save_id]
    image_path = os.path.join(PATH_TO_DATASET_IMAGES, image_name)
    
    # Read the image
    image = cv2.imread(image_path)
    if image is None:
        print(f"Warning: Could not read image {image_name}")
        continue
    
    # Read corresponding labels
    label_name = image_name.split(".")[0] + ".txt"
    label_to_read = os.path.join(PATH_TO_DATASET_LABELS, label_name)
    
    if not os.path.exists(label_to_read):
        print(f"Warning: Could not find labels for {image_name}")
        continue
    
    # Read YOLO bounding boxes
    boxes, classes = read_yolo_labels(label_to_read)
    
    # Draw bounding boxes on the image
    for box in boxes:
        x, y, w, h = yolo_to_original(image, box)
        image = cv2.rectangle(image, (x, y), (x + w, y + h), (255, 255, 255), 1)
    
    # Save annotated image
    output_image_name = "Necroptosis_" + image_name
    output_image_path = os.path.join(OUTPUT_PATH, output_image_name)
    cv2.imwrite(output_image_path, image)
    print(f"Saved: {output_image_name}")

print("✓ Processing complete!")
```
